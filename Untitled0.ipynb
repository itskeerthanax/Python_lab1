```python

#print FIZZ if a number is divisible by 3 , print BUZZ if a num is divisible
#by 5 print FIZZBUZZ if it is divisible by both 3 and 5
n = int(input())
if n%3 == 0 and n%5 == 0:
  print("FIZZBUZZ")
elif n%3 == 0:
  print("FIZZ")
elif n%5 == 0:
  print("BUZZ")
else:
  print("not divisible by 3 or 5")
```

```python
#program on "match"
day=int(input("enter day value"))
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
  case _:
    print("Invalid day")
```
