```python
#operations in tuple
t1=(1,2,3,7,10)
t2=(1,5,6)
print(t1+t2)#concatination
print(t2*4)#repetation
for i in t1:
  print(i)
print(t1[0:3])#slicing
print(t1[::-1])#reversing
print(t1[3])#indexing
```

```python
#functions in tuple
t=(1,5,6,4,8)
print(len(t))
print(max(t))
print(max(t))
print(sorted(t))

```

```python
#creating tuple
t3=tuple(map(int,input("enter tuple elements").split()))
print(t3)
```

```python
#methods of tuple
t1=(5,6,1,7,8)
t2=(1,6,7)
print(t1.index(1))
print(t2.count(1))
print(tuple(zip(t1,t2)))
```

```python
#tuple comprehension
t4=tuple(int(i) for i in input().split())
print(t4)
```
